```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv(override=True)

model_name = os.getenv('MODEL_NAME')
secondary_model = os.getenv('SECONDARY_MODEL_NAME')
base_url = os.getenv('OLLAMA_BASE_URL')

if not model_name and not base_url and not secondary_model:
  print("No MODEL_NAME or SECONDARY_MODEL_NAME or OLLAMA_BASE_URL found!")
else:
  print("MODEL_NAME, SECONDARY_MODEL_NAME and OLLAMA_BASE_URL found!, It looks good so far!")

client = OpenAI(
  base_url=base_url,
  api_key="ollama"
)
```

```python
system_message = "You are a helpful assistant"

def message_gpt(prompt):
    messages = [{"role": "system", "content": system_message}, {"role": "user", "content": prompt}]
    response = client.chat.completions.create(model=model_name, messages=messages)
    return response.choices[0].message.content
```

```python
message_gpt("what is today's date")## gives last date they trained called training cutt off
```

```python
demo = gr.Interface(fn=message_gpt, inputs="text", outputs="text")
```

```python
demo.launch()
```

```python
system_message = "You are a helpful assistant that responds in markdown without code blocks"

message_input = gr.Textbox(label="Your message:", info="Enter a message to Ai", lines=7)
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    fn=message_gpt,
    title="AI", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explain the Transformer architecture to a layperson",
        "Explain the Transformer architecture to an aspiring AI engineer",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
def stream_qwen(prompt):

  messages = [
      {"role": "system", "content": system_message},
      {"role": "user", "content": prompt}
    ]

  stream = client.chat.completions.create(
    model=model_name,
    messages=messages,
    stream=True
  )
  response = ""
  for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    yield response
```

```python
message_input = gr.Textbox(label="Your message:", info="Enter a message to qwen2.5-coder:7b", lines=7)
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    stream_qwen,
    title="AI", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explain the Transformer architecture to a layperson",
        "Explain the Transformer architecture to an aspiring AI engineer",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
def stream_llama(prompt):

  messages = [
      {"role": "system", "content": system_message},
      {"role": "user", "content": prompt}
    ]

  stream = client.chat.completions.create(
    model=secondary_model,
    messages=messages,
    stream=True
  )
  response = ""
  for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    yield response
```

```python
message_input = gr.Textbox(label="Your message:", info="llama3.1:8b", lines=7)
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    stream_qwen,
    title="AI", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explain the Transformer architecture to a layperson",
        "Explain the Transformer architecture to an aspiring AI engineer",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
def stream_model(prompt, model):
    if model=="qwen2.5-coder:7b":
        result = stream_qwen(prompt)
    elif model=="llama3.1:8b":
        result = stream_llama(prompt)
    else:
        raise ValueError("Unknown model")
    yield from result
```

```python
message_input = gr.Textbox(label="Your message:", info="Enter a message for the LLM", lines=7)
model_selector = gr.Dropdown(["qwen2.5-coder:7b", "llama3.1:8b"], label="Select model", value="qwen2.5-coder:7b")
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    stream_model,
    title="LLMs", 
    inputs=[message_input, model_selector], 
    outputs=[message_output], 
    examples=[
            ["Explain the Transformer architecture to a layperson", "qwen2.5-coder:7b"],
            ["Explain the Transformer architecture to an aspiring AI engineer", "llama3.1:8b"]
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
from scraper import fetch_website_contents
```

```python
brochure_system_prompt = """
You are an assistant that analyzes the contents of several relevant pages from a company website
and creates a short brochure about the company for prospective customers, investors and recruits.
Respond in markdown without code blocks.
Include details of company culture, customers and careers/jobs if you have the information.
"""
```

```python
def stream_brochure(company_name, url, model):

    yield ""

    prompt = f"Please generate a company brochure for {company_name}. Here is their landing page:\n"
    prompt += fetch_website_contents(url)
    if model=="qwen2.5-coder:7b":
        result = stream_qwen(prompt)
    elif model=="llama3.1:8b":
        result = stream_llama(prompt)
    else:
        raise ValueError("Unknown model")

    yield from result
```

```python
name_input = gr.Textbox(label="Company name:")
url_input = gr.Textbox(label="Landing page URL including http:// or https://")
model_selector = gr.Dropdown(["qwen2.5-coder:7b", "llama3.1:8b"], label="Select model", value="qwen2.5-coder:7b")
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    stream_brochure,
    title="LLMs", 
    inputs=[name_input,url_input,model_selector], 
    outputs=[message_output], 
    examples=[
            ["Hugging Face", "https://huggingface.co", "qwen2.5-coder:7b"],
            ["SRM MACHINE CENTER", "https://www.srmmachinecenter.com/", "llama3.1:8b"]
        ], 
    flagging_mode="never"
    )
view.launch()
```
